```python
from sklearn.preprocessing import LabelEncoder
```

```python
label = LabelEncoder()
```

```python
df_train['Stage_fear'] = label.fit_transform(df_train['Stage_fear'])
```

```python
df_test['Stage_fear'] = label.transform(df_test['Stage_fear'])
```

```python
df_train = df_train.drop('id', axis=1)
```

```python
df_test = df_test.drop('id', axis=1)
```

```python
X_train = df_train.drop('Personality', axis=1)
y_train = df_train['Personality']
x_test = df_test.copy()
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.metrics import accuracy_score
```

```python
acc_list = []
k_list = []
for k in range(1, 40):
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_val)
    acc_list.append(accuracy_score(y_val, y_pred))
    k_list.append(k)
```

```python
model = KNeighborsClassifier(n_neighbors=14)
model.fit(X_train, y_train)
y_pred = model.predict(x_test)
```

```python
df_sub['Personality'] = y_pred
```

```python
df_sub.to_csv('res.csv', index=False, index_label=False)
```
